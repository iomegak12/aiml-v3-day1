## 1. Variables and values

Use `=` to assign a value to a name. The value has a type: `int` for whole numbers, `float` for numbers with a fractional part, `str` for text, and `bool` for `True` or `False`. Use `datetime.date` for a calendar date without a time and `None` when a value is absent. Check for `None` with `is None`.

```python
from datetime import date

student_count = 24
lesson_hours = 1.5
lesson_title = "Python Essentials"
is_online = True
lesson_date = date(2026, 11, 19)
room_number = None

print(lesson_title, "on", lesson_date.isoformat())
print("Students:", student_count)
print("Duration in minutes:", lesson_hours * 60)
print("Online:", is_online)
print("Room not assigned:", room_number is None)
```

## 2. String methods

`strip()` removes whitespace from the ends, `lower()` and `upper()` change case, and `replace()` substitutes text. `split()` breaks text into a list, and `join()` combines strings using the separator that calls it. These methods leave the original string unchanged.

```python
text = "  Learn Python Today  "
clean_text = text.strip()
words = clean_text.split()

print("Trimmed:", clean_text)
print("Lowercase:", clean_text.lower())
print("Uppercase:", clean_text.upper())
print("Replaced:", clean_text.replace("Today", "Together"))
print("Words:", words)
print("Joined:", "-".join(words))
```

## 3. Lists, tuples, sets, and dictionaries

- A **list** is an ordered collection you can change.
- A **tuple** is an ordered collection whose items cannot be reassigned.
- A **set** stores unique items without a guaranteed order. This example sorts the items before printing.
- A **dictionary** maps keys to values.

```python
colors = ["blue", "green"]
colors.append("red")
print("First color:", colors[0], "| All colors:", colors)

point = (3, 5)
x, y = point
print("Coordinates:", x, y)

unique_tags = set(["python", "basics", "python"])
print("Unique tags:", sorted(unique_tags))

stock = {"pens": 12, "pencils": 8}
stock["pens"] += 2
print("Pens in stock:", stock["pens"])
```

## 4. List comprehensions

A list comprehension builds a new list. Compare a `for` loop and a comprehension that both square each number. Then add an `if` condition to keep only even numbers before squaring them.

```python
numbers = [1, 2, 3, 4, 5, 6]

squares_from_loop = []
for number in numbers:
    squares_from_loop.append(number ** 2)

squares = [number ** 2 for number in numbers]
even_squares = [number ** 2 for number in numbers if number % 2 == 0]

print("From the loop:", squares_from_loop)
print("From the comprehension:", squares)
print("Squares of even numbers:", even_squares)
```

## 5. Default and named arguments

A parameter takes its default value when the caller leaves it out. Use `name=value` in a call to pass a named (keyword) argument; named arguments do not have to follow the order of the parameters.

```python
def greet(name, greeting="Hello"):
    return f"{greeting}, {name}!"


print(greet("Anu"))
print(greet(greeting="Good morning", name="Anu"))
```

## 6. Handling errors

`try` wraps code that may raise an exception. `except ValueError` handles that specific error, and `finally` runs whether the conversion succeeds or fails.

```python
for entry in ["42", "forty-two"]:
    try:
        number = int(entry)
        print("Converted:", number)
    except ValueError:
        print(f"Cannot convert {entry!r} to an integer.")
    finally:
        print(f"Finished processing {entry!r}.")
```

## 7. Reading and appending a text file

Use `w` to create the sample file, `r` to read it, and `a` to add text without replacing what is already there. Each `with open(...)` block closes the file automatically. The temporary directory and its file are removed when the outer `with` block ends.

```python
from pathlib import Path
from tempfile import TemporaryDirectory

with TemporaryDirectory() as temp_dir:
    notes_path = Path(temp_dir) / "notes.txt"

    with open(notes_path, "w", encoding="utf-8") as notes:
        notes.write("First note\n")

    with open(notes_path, "r", encoding="utf-8") as notes:
        print("Before appending:")
        print(notes.read(), end="")

    with open(notes_path, "a", encoding="utf-8") as notes:
        notes.write("Second note\n")

    with open(notes_path, "r", encoding="utf-8") as notes:
        print("After appending:")
        print(notes.read(), end="")
```

## 8. A simple class

`__init__` sets up the new object by assigning its attributes. `__str__` returns a readable string for that object, which `print()` uses.

```python
class Book:
    def __init__(self, title, pages):
        self.title = title
        self.pages = pages

    def __str__(self):
        return f"{self.title} ({self.pages} pages)"


book = Book("Everyday Python", 120)
print(book)
```
